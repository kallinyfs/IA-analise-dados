```python
import seaborn as sns
```

```python
grafico = sns.histplot(data=df, x="categoria", kde=True, bins=5)
grafico.set_title("Produtos mais vendidos")
grafico.set_xlabel("Categorias")
grafico.set_ylabel("Quantidade vendida")
```

```python
grafico = sns.boxplot(data=df, x="categoria", y="preco_unitario")
grafico.set_title("Preço por Categoria")
grafico.set_xlabel("Podutos")
grafico.set_ylabel("Preço Unitário")
```
