17- **Criação dos principais gráficos:**

```python
sns.countplot(data=df, y="produto")
plt.title("Produtos mais vendidios")
```

18- Primeiro transforma a coluna data em tipo datetime, depois extrai apenas o mês da coluna de data, po último agrupa os meses e o total e soma o valor por mês:

```python
df["data"] = pd.to_datetime(df["data"])
df["mes"] = df["data"].dt.month
df.groupby("mes")["total"].sum()
```

Avaliação por canal

```python
df.groupby("canal")["avaliacao"].mean()
```

Tempo de espera por canal

```python
df.groupby("canal")["tempo_espera_min"].mean()
```

**Faturamento por produto**

```python
faturamento_por_produto = df.groupby("produto")["total"].sum().reset_index()
sns.barplot(data=faturamento_por_produto, x="total", y="produto")
plt.show()
```

**Faturamento por dia da semana**

```python
faturamento_dia = df.groupby("dia_semana")["total"].sum().reset_index()
sns.barplot(data=faturamento_dia, x="total", y="dia_semana")
plt.show()
```

**Mapa de Calor**

round = arredondar
annot=True: mostra as informações numéricas

```python
numeros = df[["quantidade", "preco_unitario", "tempo_espera_min", "avaliacao", "total", "mes"]]
sns.heatmap(numeros.corr().round(2), annot=True)
```
