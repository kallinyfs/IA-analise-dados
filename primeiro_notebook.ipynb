```python
import seaborn as sns
```

```python
#Quanto mais bins mais colunas
#mostra como as temperaturas cresceram
#set.title: adiciona titulo
#set.xlabel: nemeia o eixo x
#set.ylabel: nemeia o eixo y

grafico = sns.histplot(data=df, x="temperatura_c", kde=True, bins=20)
grafico.set_title("Distribuição de Temperatura")
grafico.set_xlabel("Temperatura em °C")
grafico.set_ylabel("Registros")
```
